```python
import asyncio
import os
from telegram import Bot
import dotenv
CHAT_ID = 8642372237
dotenv.load_dotenv()
TELEGRAM_BOT_TOKEN = os.getenv("TELEGRAM_BOT_TOKEN")
```

```python
async def main() -> None:
    bot = Bot(TELEGRAM_BOT_TOKEN)

    async with bot:
        await bot.send_message(
            chat_id=CHAT_ID,
            text="테스트",
        )


await main()
```
