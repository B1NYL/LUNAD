```python
import asyncio
import os
from telegram import Bot
import dotenv
CHAT_ID = 7594336167
dotenv.load_dotenv()
```

```python
async def main() -> None:
    bot = Bot(os.getenv("TELEGRAM_BOT_TOKEN"))          
    async with bot:
        await bot.send_message(
            chat_id=CHAT_ID,
            text="테스트",
        )

await main()
```
